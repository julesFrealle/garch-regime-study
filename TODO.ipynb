$\textbf{Un nouveau critère de validation pour la VaR}$:\
On peut définir la VaR par $VaR_{t} = \arg\min E\big((1-\alpha)(\epsilon_t-VaR_t)*1_{\epsilon_t-VaR_t>0} + \alpha(VaR_t-\epsilon_t)*1_{Var_t-\epsilon_t>0}\big)$
Ainsi un critère peut être $\hat{VaR}_{t,model 1}<\hat{VaR}_{t,model 2}$ avec $\hat{}$ où on remplace l'espérance par la moyenne.\
On peut ensuite faire un test de Dieblod-Mariano pour la significativité de la différence.

$\textbf{Comparaison de }\sigma_t^2$ et $\epsilon_t^2$:\
Un bon critère est $\frac{\epsilon_t^2}{\sigma_t^2}+\log\sigma_t^2$, il est minimisé par $E(\epsilon_t^2)$ et a une meilleure gueule qu'une MSE qui serait sensible aux outliers.\
